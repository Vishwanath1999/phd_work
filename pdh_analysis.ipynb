```python
def process_signals(det_jitter, power_response, window_size=7):
    window = window_size # small window for local mean
    det_dither = det_jitter - np.convolve(det_jitter, np.ones(window)/window, mode='same') # analogous to high-pass filter
    power_dither = power_response - np.convolve(power_response, np.ones(window)/(window), mode='same') # analogous to high-pass filter

    # 3. Synchronous Detection (The Mixer)
    # Multiplying the input jitter by the power response reveals the 'direction'
    error_signal_raw = det_dither * power_dither

    # 4. Low-pass filter the product to get the DC Error Signal (The S-Curve)
    error_signal = np.convolve(error_signal_raw, np.ones(50)/50, mode='same')
    return error_signal

def process_signals_fir(
    det_jitter,
    power_response,
    *,
    fs_hz,
    hp_cutoff_hz,
    lp_cutoff_hz,
    numtaps_hp=401,
    numtaps_lp=401,
    window=("kaiser", 8.0),
):
    """
    1) FIR high-pass both signals to isolate fast 'dither'
    2) Multiply (mixer)
    3) FIR low-pass the product to extract DC error
    """
    det_jitter = np.asarray(det_jitter, dtype=float)
    power_response = np.asarray(power_response, dtype=float)
    if det_jitter.shape != power_response.shape:
        raise ValueError("det_jitter and power_response must have the same shape.")

    nyq = 0.5 * fs_hz
    if not (0 < hp_cutoff_hz < nyq) or not (0 < lp_cutoff_hz < nyq):
        raise ValueError("Cutoffs must satisfy 0 < cutoff < Nyquist.")

    # FIR HP: remove slow drift
    b_hp = firwin(
        numtaps=numtaps_hp,
        cutoff=hp_cutoff_hz,
        pass_zero=False,   # high-pass
        fs=fs_hz,
        window=window,
    )

    det_dither = filtfilt(b_hp, [1.0], det_jitter)
    power_dither = filtfilt(b_hp, [1.0], power_response)

    # Mixer
    error_signal_raw = det_dither * power_dither

    # FIR LP: extract DC component after mixing
    b_lp = firwin(
        numtaps=numtaps_lp,
        cutoff=lp_cutoff_hz,
        pass_zero=True,    # low-pass
        fs=fs_hz,
        window=window,
    )
    error_signal = filtfilt(b_lp, [1.0], error_signal_raw)

    return error_signal

def plot_all(det_jitter, eff_detuning, thermal_shift, power_response, error_signal, name=None, title=None, save=False):
    # --- PRE-CALCULATION FOR STIFFNESS ---
    # 1. Focus only on the 'locked' phase (last 20% of the data) to get a clean slope
    lock_start = int(len(eff_detuning) * 0.6)
    x_lock = eff_detuning[lock_start:-100]
    y_lock = error_signal[lock_start:-100]

    # 2. Linear regression to find the slope (Stiffness)
    # This finds the 'm' in y = mx + c near the zero-crossing
    slope, intercept = np.polyfit(x_lock, y_lock, 1)
    stiffness = np.abs(slope)

    # --- PLOTTING ---
    fig = plt.figure(figsize=(14, 10), constrained_layout=True)
    gs = fig.add_gridspec(nrows=3, ncols=2, height_ratios=[1, 1, 2])

    ax1 = fig.add_subplot(gs[0, 0])
    ax2 = fig.add_subplot(gs[0, 1])
    ax3 = fig.add_subplot(gs[1, 0])
    ax4 = fig.add_subplot(gs[1, 1])
    ax5 = fig.add_subplot(gs[2, :])

    # 1) Detuning jitter
    ax1.plot(det_jitter)
    ax1.set_xlabel('Time step')
    ax1.set_ylabel('Detuning Jitter (GHz)')
    ax1.grid(True)
    ax1.set_title('Detuning Jitter over Time')

    # 2) Effective detuning
    ax2.plot(eff_detuning)
    ax2.set_xlabel('Time step')
    ax2.set_ylabel('Effective Detuning (GHz)')
    ax2.grid(True)
    ax2.set_title('Effective Detuning over Time')

    # 3) Thermal shift
    ax3.plot(thermal_shift)
    ax3.set_xlabel('Time step')
    ax3.set_ylabel('Thermal Shift (GHz)')
    ax3.grid(True)
    # ax3.set_ylim(-6.4, -6)
    ax3.set_title('Thermal Shift over Time')

    # 4) Cavity power
    ax4.plot(power_response)
    ax4.set_yscale('log')
    ax4.set_xlabel('Time step')
    ax4.set_ylabel('Cavity Power (mW)')
    ax4.grid(True)
    ax4.set_title('Cavity Power Response over Time')

    # 5) Error signal (S-Curve) with Stiffness Display
    scatter = ax5.scatter(
        eff_detuning[100:-100],
        error_signal[100:-100],
        c=np.arange(len(error_signal[100:-100])),
        cmap='viridis',
        s=20,
        alpha=0.5
    )

    # Plot the fitted slope line near the crossing for visual confirmation
    x_range = np.linspace(np.min(x_lock), np.max(x_lock), 100)
    ax5.plot(x_range, slope * x_range + intercept, color='red', linewidth=2, label='Lock Slope')

    ax5.axhline(0, color='black', linestyle='--')
    ax5.set_xlabel('Effective Detuning ($\\Delta f_{eff}$)', fontsize=18)
    ax5.grid(True)
    ax5.set_ylabel('Extracted Error Signal (a.u.)', fontsize=18)
    ax5.set_title(f'PDH S-Curve | Locking Stiffness: {stiffness:.2e} a.u./GHz', fontsize=20)
    ax5.tick_params(axis='both', labelsize=14)
    ax5.legend(fontsize=14)
    # ax5.set_xlim(-5,-2.5)
    # ax5.set_ylim(-0.2,0.2)

    cbar = fig.colorbar(scatter, ax=ax5)
    cbar.set_label('Time Step', fontsize=14)
    # set title for figure
    if title is not None:
        fig.suptitle(title, fontsize=22)
        plt.subplots_adjust(top=0.92)
    # if save is true save as pdf and png
    if save == True:
        plt.savefig(f'{name}.pdf')
        plt.savefig(f'{name}.png', dpi=300)

    plt.show()

from scipy.fft import fft, fftfreq

# def get_ringing_freq(power_data, dt):
#     # Use the 'locked' portion of the data
#     data = power_data[-int(len(power_data)*0.4):]
#     # Remove DC component (the average power)
#     data_ac = data - np.mean(data)
    
#     n = len(data_ac)
#     # Compute FFT
#     yf = fft(data_ac)
#     xf = fftfreq(n, dt)
    
#     # Get magnitude and find the peak
#     magnitude = 2.0/n * np.abs(yf[0:n//2])
#     frequencies = xf[0:n//2]
    
#     peak_idx = np.argmax(magnitude[1:]) + 1 # ignore DC
#     return frequencies, magnitude, frequencies[peak_idx]

# def calc_gain_phase_margin(power_data, ctrl_freq):
# # Constants from your setup
#     tR = 1e-12 # 1 ps
#     ctrl_delay = ctrl_freq * tR # 0.5 ns
#     ctrl_scale = ctrl_freq/100

#     # Calculate for 5x case
#     freqs, mag, ringing_f = get_ringing_freq(power_data, ctrl_delay)
#     phase_margin = 90 - 360 * ringing_f * ctrl_delay
#     gain_margin = 20*np.log10(90/(90 - phase_margin))
#     print(f"{ctrl_scale:.1f}x Ctrl Delay: Ringing Frequency = {ringing_f*1e-6:.2f} MHz, Phase Margin = {phase_margin:.2f} degrees, Gain Margin = {gain_margin:.2f} dB")

#     plt.figure(figsize=(10, 5))
#     plt.loglog(freqs, mag, label=f'{ctrl_scale:.1f}x') # Plot in GHz
#     plt.axvline(ringing_f , color='r', linestyle='--', label=f'Ringing 5x: {ringing_f*1e-6:.2f} MHz')

#     plt.xlabel('Frequency (Hz)', fontsize=16)
#     plt.ylabel('Magnitude', fontsize=16)
#     plt.title('Power Response FFT: Identifying Phase Margin Ringing', fontsize=18)
#     plt.xticks(fontsize=14)
#     plt.yticks(fontsize=14)
#     plt.grid(True, which='both', ls='--', lw=0.5, axis='both')
#     # write the gain and phase margin in the plots itself
#     plt.text(0.02, 0.08, f'{ctrl_scale:.1f}x Ctrl Delay:\nPhase Margin: {phase_margin:.2f}°\nGain Margin: {gain_margin:.2f} dB', transform=plt.gca().transAxes, fontsize=12, bbox=dict(facecolor='white', alpha=0.7))
#     plt.legend(fontsize=12)
#     plt.show()

def get_ringing_freq(power_data, dt, freq_limit=1e9):
    # Use the 'locked' portion of the data (last 40%)
    data = power_data[-int(len(power_data)*0.55):]
    # subtract windowes mean to remove DC component
    data_ac = data#- np.convolve(data, np.ones(500)/500, mode='same')
    # data_ac = data - np.mean(data)
    
    n = len(data_ac)
    yf = fft(data_ac)
    xf = fftfreq(n, dt)
    
    # Get positive frequencies and magnitude
    full_freqs = xf[0:n//2]
    full_mag = 2.0/n * np.abs(yf[0:n//2])
    
    # --- NEW: Frequency Masking ---
    # We ignore artifacts above 1 GHz to find the 'True' control loop peak
    mask = full_freqs <= freq_limit
    frequencies = full_freqs[mask]
    magnitude = full_mag[mask]
    
    # Identify peak within the masked range (ignoring DC)
    peak_idx = np.argmax(magnitude[1:]) + 1 
    return frequencies, magnitude, frequencies[peak_idx]

def calc_gain_phase_margin(power_data, ctrl_freq, name=None, save=False):
    tR = 1e-12 
    ctrl_delay = ctrl_freq * tR 
    ctrl_scale = ctrl_freq / 100

    # Calculate metrics within the 1GHz limit
    freqs, mag, ringing_f = get_ringing_freq(power_data, ctrl_delay, freq_limit=1e9)
    
    # Phase Margin formula: 90 - (delay_contribution)
    phase_margin = 90 - 360 * ringing_f * ctrl_delay
    
    # Gain Margin formula: 20*log10(90 / (90 - PM))
    # Handling cases where PM is very close to 90 to avoid division by zero
    denom = max(90 - phase_margin, 1e-6)
    gain_margin = 20 * np.log10(90 / denom)
    
    print(f"{ctrl_scale:.1f}x Ctrl Delay: Ringing Frequency = {ringing_f*1e-6:.2f} MHz, "
          f"Phase Margin = {phase_margin:.2f} degrees, Gain Margin = {gain_margin:.2f} dB")

    # Plotting
    plt.figure(figsize=(10, 5))
    plt.loglog(freqs, mag, label=f'{ctrl_scale:.1f}x Scale')
    plt.axvline(ringing_f, color='r', linestyle='--', label=f'Ringing Peak: {ringing_f*1e-6:.2f} MHz')

    plt.xlabel('Frequency (Hz)', fontsize=16)
    plt.ylabel('Magnitude', fontsize=16)
    plt.title('Power Response FFT: True Phase Margin Analysis', fontsize=18)
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    plt.grid(True, which='both', ls='--', lw=0.5)
    
    # Informative text box
    info_str = (f'{ctrl_scale:.1f}x Ctrl Delay\n'
                f'Phase Margin: {phase_margin:.2f}°\n'
                f'Gain Margin: {gain_margin:.2f} dB')
    plt.text(0.05, 0.05, info_str, transform=plt.gca().transAxes, 
             fontsize=12, verticalalignment='bottom', bbox=dict(facecolor='white', alpha=0.8))
    
    plt.legend(fontsize=12)
    plt.tight_layout()
    if save==True:
        plt.savefig(f'{name}_gain_phase_margin.pdf')
        plt.savefig(f'{name}_gain_phase_margin.png', dpi=300)
    plt.show()
```
